```python
from datasets import load_dataset
import json
 
# Login using e.g. `huggingface-cli login` to access this dataset
ds = load_dataset("UBC-NLP/palm")
 
# Filter dataset for country='Tunisia' and question_type='open-ended'
filtered_ds = ds.filter(lambda x: x['country'] == 'Tunisia' and x['question_type'] == 'open-ended')
 
# Convert to JSON serializable format and save to file
output_path = "C:\\Users\\user\\OneDrive\\Bureau\\TunisianDialogSystem\\llm\\sft\\data\\raw\\filtered_data1.json"
filtered_data_list = [example for example in filtered_ds["train"]]
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(filtered_data_list, f, ensure_ascii=False, indent=4)
```

```python
# Count the number of questions in the filtered dataset
num_questions = len(filtered_data_list)
print(f"Number of questions: {num_questions}")
```
